```python
limit = 1
page = 1

def url_builder(): # I can put url checker here
    return f"https://api.darkerdb.com/v1/items?key={os.getenv("dark_api_key")}&limit={limit}&page={page}"

if requests.get(url_builder()).status_code == 200:
    with requests.Session() as s:   
        table = pd.DataFrame()
        byte_counter = 0
        json = {"pagination": {"count": int(limit)}} # reset the loop
        while json['pagination']['count'] != 0:
            url = url_builder()
            request = s.get(url)
            json = request.json()
            pandas = pd.json_normalize(json, record_path='body', sep=',')
            table = pd.concat([table, pandas], ignore_index=True)
            page = str(int(page) + 1)
            byte_counter += len(request.content) # 8 bits in a byte
    print(request.status_code)
                        
else:
    error_code = requests.get(url_builder()).status_code
    print(f"Status Code{error_code}: {HTTPStatus(error_code).phrase} | {HTTPStatus(error_code).description}" )
```
