```python
# standard
import os

# third party
import requests
from http import HTTPStatus
import pandas as pd
```

```python
# Real time pricing
item = 'Dashing Boots'
rarity = 'Epic'
secondary1 = "[dexterity]=2"
secondary2 = " "
secondary3 = " "


limit = '50'
page = '1'


url = f"https://api.darkerdb.com/v1/market?key={os.getenv("dark_api_key")}&item={item}&rarity={rarity}&secondary{secondary1}&secondary{secondary2}&secondary{secondary3}&condense=true&limit={limit}&page={page}"


request = requests.get(url)

if request.status_code == 202:
    json = request.json()
    pandas = pd.json_normalize(json, record_path='body', sep=',')
    print(f"Status Code{request.status_code}: {HTTPStatus(request.status_code).phrase} | {HTTPStatus(request.status_code).description}" )
else:
    print(f"Status Code{request.status_code}: {HTTPStatus(request.status_code).phrase} | {HTTPStatus(request.status_code).description}" )
```
