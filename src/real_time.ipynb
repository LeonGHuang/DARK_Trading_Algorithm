```python
# standard

# third party
import requests
from http import HTTPStatus
import pandas as pd

# local
import key
```

```python
# Real time pricing
item = 'Dashing Boots'
rarity = 'Epic'
secondary1 = "[dexterity]=2"
secondary2 = " "
secondary3 = " "


limit = '50'
page = '1'


url = f"https://api.darkerdb.com/v1/market?key={key.API}&item={item}&rarity={rarity}&secondary{secondary1}&secondary{secondary2}&secondary{secondary3}&condense=true&limit={limit}&page={page}"


request = requests.get(url)

if request.status_code == 202:
    json = request.json()
    pandas = pd.json_normalize(json, record_path='body', sep=',')
else:
    print(HTTPStatus(request.status_code).description)
```
