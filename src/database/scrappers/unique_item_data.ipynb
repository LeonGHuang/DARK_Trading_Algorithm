```python
import requests
import os
import pandas as pd
```

```python
def url(page):
    url = f"https://api.darkerdb.com/v1/items?key={os.getenv("dark_api_key")}&limit=50&page={page}"
    return url
```

```python
with requests.Session() as ses:
    req = ses.get(url('1'))
    body_list = []

    request_count = 0
    for page in range(1,req.json()['pagination']['num_pages']):
        body_list.extend(req.json()['body'])
        req = ses.get(req.json()['pagination']['next'])

        request_count += 1
        print(request_count, end="\r")
```
