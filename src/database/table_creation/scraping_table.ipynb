```python
import src.database.scripts.sql as sql
```

```python
def drop_table(cursor):
    query = "DROP TABLE IF EXISTS scraping"
    cursor.execute(query)


def create_table(cursor):
    query = """
    CREATE TABLE IF NOT EXISTS scraping (
        id              TEXT,
        rarity          TEXT,
        cost_per_slot   INT
    CONSTRAINT      fk_id
    FOREIGN KEY     (id)
    REFERENCES      item_data(id)
    )"""
    cursor.execute(query)


def insert_values(cursor):
    val = [
        ('WorkshopScraps_3001', 'Uncommon', 2), 
        ('WorkshopScraps_4001', 'Rare', 3),    
        ('WorkshopScraps_5001', 'Epic', 5),    
        ('WorkshopScraps_6001', 'Legenedary', 7),    
        ('WorkshopScraps_7001', 'Unique', 10)
        ]

    query = """
    INSERT INTO scraping (id, rarity, cost_per_slot)
    VALUES (%s, %s, %s)
    """
    cursor.executemany(query, val)


def main():
    conn = sql.connect_pc()
    cursor = conn.cursor()

    print("dropping table...")
    drop_table(cursor)
    print("creating table...")
    create_table(cursor)
    print("inserting table...")
    insert_values(cursor)
```

```python
main()
```
