```python
# Creating a function for market

table = pd.DataFrame()
byte_counter = 0


item = 'Gold Coin Chest'
rarity = 'Unique'

start = "2025-11-15T00:00:00Z" #UTC Time
end = "2025-11-16T24:00:00Z"

has_sold = "1"

limit = '50'
page = '1'
json = {"pagination": {"count": int(limit)}} # initiate/reset for loop

while json['pagination']['count'] == int(limit):
    url = f"https://api.darkerdb.com/v1/market?key={API.key}&item={item}&rarity={rarity}&from={start}&to={end}&has_sold={has_sold}&condense=true&limit={limit}&page={page}"

    request = requests.get(url)
    json = request.json()
    pandas = pd.json_normalize(json, record_path='body', sep=',')
    table = pd.concat([table, pandas], ignore_index=True)

    page = str(int(page) + 1)
    byte_counter += len(request.content)

print(f"request = {page}")
print(f"bytes = {byte_counter}")

```

```python
# to_sql doesnt support priumary key
# SQLite doesnt allow contraint to be altered in

query = f'''
CREATE TABLE IF NOT EXISTS "{item}" (
    id INTEGER INTEGER,
    cursor INTEGER PRIMARY KEY,
    item_id TEXT,
    item TEXT,
    archetype TEXT,
    rarity TEXT,
    price INTEGER,
    price_per_unit TEXT,
    quantity INTEGER,
    created_at TEXT,
    expires_at TEXT,
    sold_at TEXT,
    has_sold INTEGER,
    has_expired INTEGER,
    seller TEXT
    )'''

cursor.execute(query)
```
